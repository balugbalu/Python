```python
import rawpy
import imageio
from PIL import Image
import os

def convert_arw_to_jpg(source_dir, output_dir, quality=85):
    # Create output directory if it doesn't exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    for filename in os.listdir(source_dir):
        if filename.lower().endswith(".arw"):
            print(f"Processing {filename}...")
            
            file_path = os.path.join(source_dir, filename)
            
            try:
                # Use rawpy to read the ARW file
                with rawpy.imread(file_path) as raw:
                    # Postprocess converts the raw data into an RGB array
                    # use_camera_wb=True keeps the colors as shot
                    rgb = raw.postprocess(use_camera_wb=True, no_auto_bright=False, bright=1.0)
                
                # Convert the numpy array to a Pillow Image object
                img = Image.fromarray(rgb)
                
                # Define output path (change extension to .jpg)
                target_name = os.path.splitext(filename)[0] + ".jpg"
                target_path = os.path.join(output_dir, target_name)
                
                # Save with compression (quality 1-95, 85 is a good sweet spot)
                img.save(target_path, "JPEG", quality=quality, optimize=True)
                
            except Exception as e:
                print(f"Failed to convert {filename}: {e}")

# Usage
source_folder = r"C:\Users\bkgha\OneDrive\Pictures\sow"
destination_folder = r"C:\Users\bkgha\OneDrive\Pictures\sow\output"

convert_arw_to_jpg(source_folder, destination_folder, quality=80)
print("Conversion complete!")
```
